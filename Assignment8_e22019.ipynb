###Q. Statistical Mechanics of the Quantum Oscillator

#### 1. The Partition Function

$$E_j = \hbar\omega\left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$
$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{E_j}{k_B T}}$$
Letting $\beta = \frac{1}{k_B T}$, $$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar\omega\left(j + \frac{1}{2}\right)} = \sum_{j=0}^{\infty} e^{-\frac{1}{2}\beta\hbar\omega} e^{-j\beta\hbar\omega}$$
$$Z(T) = e^{-\frac{1}{2}\beta\hbar\omega} \sum_{j=0}^{\infty} \left(e^{-\beta\hbar\omega}\right)^j$$
The summation is an infinite geometric series of the form $\sum_{j=0}^{\infty} r^j$ with a common ratio $r = e^{-\beta\hbar\omega}$. Because $\hbar, \omega, T > 0$, the exponent $\beta\hbar\omega > 0$, meaning the common ratio satisfies $0 < r < 1$. Thus, the geometric series safely converges to $\frac{1}{1-r}$:$$\sum_{j=0}^{\infty} \left(e^{-\beta\hbar\omega}\right)^j = \frac{1}{1 - e^{-\beta\hbar\omega}}$$

### 2. Thermodynamic Potentials
$$Z(T) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

#### A. Helmholtz Free Energy, $F(T)$

 $$F(T) = -k_B T \ln Z$$ $$\ln Z = \ln\left( \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right) = \ln\left(e^{-\frac{\hbar\omega}{2k_B T}}\right) - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$$$\ln Z = -\frac{\hbar\omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$
$$F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

#### B. Internal Energy, $U(T)$

$$U(T) = -\frac{\partial \ln Z}{\partial \beta}$$
$\beta = \frac{1}{k_B T}$
$$\ln Z = -\frac{1}{2}\beta\hbar\omega - \ln\left(1 - e^{-\beta\hbar\omega}\right)$$
$$\frac{\partial \ln Z}{\partial \beta} = -\frac{1}{2}\hbar\omega - \frac{1}{1 - e^{-\beta\hbar\omega}} \cdot \left(-e^{-\beta\hbar\omega}\right) \cdot (-\hbar\omega)$$$$\frac{\partial \ln Z}{\partial \beta} = -\frac{1}{2}\hbar\omega - \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}$$
$$\frac{\partial \ln Z}{\partial \beta} = -\frac{1}{2}\hbar\omega - \frac{\hbar\omega}{e^{\beta\hbar\omega} - 1}$$
$$U(T) = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right)$$


### 3. Entropy

 $$S = -\left(\frac{\partial F}{\partial T}\right)$$ $$\frac{\partial F}{\partial T} = 0 + k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + k_B T \cdot \left[ \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar\omega}{k_B T}}\right) \cdot \left(\frac{\hbar\omega}{k_B T^2}\right) \right]$$$$\frac{\partial F}{\partial T} = k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - \frac{\hbar\omega}{T} \frac{e^{-\frac{\hbar\omega}{k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$
$$\frac{\partial F}{\partial T} = k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

#### Verification of $U = F + TS$

 $$F + TS$$
 $$F + TS = \left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right]$$$$F + TS = \frac{\hbar\omega}{2} + {k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)} - {k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)} + \hbar\omega \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$$$F + TS = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right) = U(T)$$
The relationship $U = F + TS$ is rigorously verified.

### 4. Heat Capacity and Hyperbolic Form

#### A. Calculating $C_V$

$$C_V = \left(\frac{\partial U}{\partial T}\right)$$
$$C_V = \frac{\partial}{\partial T} \left[ \frac{\hbar\omega}{2} + \hbar\omega \left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-1} \right]$$$$C_V = \hbar\omega \cdot (-1)\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-2} \cdot \left(e^{\frac{\hbar\omega}{k_B T}}\right) \cdot \left(-\frac{\hbar\omega}{k_B T^2}\right)$$$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2}$$

#### B. Proof of Hyperbolic Form

 $$x = \frac{\hbar\omega}{2k_B T}$$
 $$C_V = k_B (2x)^2 \frac{e^{2x}}{(e^{2x} - 1)^2} = 4k_B x^2 \frac{e^{2x}}{(e^{2x} - 1)^2}$$
$$\frac{e^{2x} \cdot e^{-2x}}{(e^{2x} - 1)^2 \cdot e^{-2x}} = \frac{1}{\left[(e^{2x} - 1)e^{-x}\right]^2} = \frac{1}{(e^x - e^{-x})^2}$$
Using the given hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$, $$(e^x - e^{-x})^2 = (2\sinh(x))^2 = 4\sinh^2(x)$$
$$C_V = 4k_B x^2 \cdot \frac{1}{4\sinh^2(x)} = k_B x^2 \frac{1}{\sinh^2(x)}$$
$$C_V = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

### 5. Asymptotic Behavior

In the high-temperature limit ($k_B T \gg \hbar\omega$), the parameter $x = \frac{\hbar\omega}{2k_B T}$ becomes extremely small ($x \to 0$).
We can evaluate the behavior of $\sinh(x)$ using its Taylor series expansion for small inputs:$$\sinh(x) \approx x + \frac{x^3}{6} + \dots \approx x$$
Squaring the leading term gives $\sinh^2(x) \approx x^2$. Substituting this approximation directly into our compact hyperbolic heat capacity formula:$$C_V \approx k_B x^2 \left( \frac{1}{x^2} \right) = k_B$$

#### Match with the Classical Equipartition Theorem:

Yes, this perfectly matches the prediction of the classical equipartition theorem.
According to classical thermodynamics, every independent quadratic degree of freedom contributes $\frac{1}{2}k_B$ to a system's heat capacity. A classical one-dimensional harmonic oscillator has exactly two quadratic degrees of freedom:

*   Kinetic Energy: $K = \frac{p^2}{2m}$
*   Potential Energy: $V = \frac{1}{2}m\omega^2 q^2$

Therefore, classical mechanics dictates that $C_V = 2 \times \left(\frac{1}{2}k_B\right) = k_B$. At high temperatures, thermal fluctuations ($k_B T$) drastically dwarf the discrete energy gaps ($\hbar\omega$). As a result, the quantum granularity of the energy states becomes completely blurred out, causing the quantum oscillator to converge beautifully back into its smooth, classical continuum limit.

# Maximum Entropy Distribution (Gaussian Derivation)

### 1. The Continuous Entropy Functional

$$H_c[p] \triangleq -\int_{-\infty}^{\infty} p(x) \ln\left(\frac{p(x)}{m(x)}\right)dx$$
By choosing a completely uniform baseline reference measure $m(x) = 1$ $$H[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx$$



*   Total Probability Axiom (Normalization):$$\int_{-\infty}^{\infty} p(x) \, dx = 1$$
*   The Mean Constraint:$$\mathbb{E}[X] = \int_{-\infty}^{\infty} x \cdot p(x) \, dx = \mu$$
*   The Variance Constraint:$$\mathbb{E}[(X-\mu)^2] = \int_{-\infty}^{\infty} (x-\mu)^2 \cdot p(x) \, dx = \sigma^2$$

$$\mathcal{L}[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \gamma \left( \int_{-\infty}^{\infty} p(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x p(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx - \sigma^2 \right)$$
Where $\gamma$, $\lambda_1$, and $\lambda_2$ are the corresponding Lagrange multipliers.

$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \gamma - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$
$$\ln p(x) = -(1 + \gamma) - \lambda_1 x - \lambda_2 (x-\mu)^2$$
$$p^*(x) = \exp(-(1+\gamma)) \exp\left(-\lambda_1 x - \lambda_2 (x-\mu)^2\right)$$
By defining the continuous Partition Function (normalization constant) as $Z \triangleq \exp(1+\gamma)$, the optimal probability density function takes the generalized form:$$p^*(x) = \frac{1}{Z} \exp\left(-\lambda_1 x - \lambda_2 (x-\mu)^2\right)$$

Proving that $\lambda_1 = 0$

Let us expand the exponent algebraically to analyze its structural properties:$$-\lambda_1 x - \lambda_2 (x-\mu)^2 = -\lambda_2 \left(x^2 - 2\mu x + \mu^2\right) - \lambda_1 x$$$$= -\lambda_2 x^2 + (2\lambda_2 \mu - \lambda_1)x - \lambda_2 \mu^2$$
This matches the general algebraic quadratic form of a Gaussian distribution, $\exp(-Ax^2 + Bx + C)$. The expected value (mean) of a distribution proportional to $\exp(-Ax^2 + Bx)$ is given analytically by $\frac{B}{2A}$. $A = \lambda_2$ and $B = 2\lambda_2 \mu - \lambda_1$, $$\mathbb{E}[X] = \frac{2\lambda_2 \mu - \lambda_1}{2\lambda_2} = \mu - \frac{\lambda_1}{2\lambda_2}$$
Because our mean constraint explicitly mandates that $\mathbb{E}[X] = \mu$, it forces:$$\mu - \frac{\lambda_1}{2\lambda_2} = \mu \implies \lambda_1 = 0$$
$$p^*(x) = \frac{1}{Z} \exp\left(-\lambda_2 (x-\mu)^2\right)$$



$$\int_{-\infty}^{\infty} p^*(x) \, dx = \frac{1}{Z} \int_{-\infty}^{\infty} \exp\left(-\lambda_2 (x-\mu)^2\right) dx = 1$$$$Z = \int_{-\infty}^{\infty} \exp\left(-\lambda_2 (x-\mu)^2\right) dx$$
$$Z = \int_{-\infty}^{\infty} \exp(-\lambda_2 u^2) du = \sqrt{\frac{\pi}{\lambda_2}}$$

$$\sigma^2 = \int_{-\infty}^{\infty} (x-\mu)^2 \cdot p^*(x) \, dx = \frac{1}{Z} \int_{-\infty}^{\infty} u^2 \exp(-\lambda_2 u^2) du$$
$$\int_{-\infty}^{\infty} u^2 \exp(-\lambda_2 u^2) du = \frac{1}{2} \sqrt{\frac{\pi}{\lambda_2^3}}$$
$$\sigma^2 = \frac{1}{\sqrt{\frac{\pi}{\lambda_2}}} \cdot \left( \frac{1}{2} \sqrt{\frac{\pi}{\lambda_2^3}} \right) = \frac{1}{2\lambda_2}$$
$$\lambda_2 = \frac{1}{2\sigma^2}$$

$$Z = \sqrt{\frac{\pi}{1/(2\sigma^2)}} = \sqrt{2\pi\sigma^2}$$


Substituting the derived parameters ($\lambda_1 = 0$, $\lambda_2 = \frac{1}{2\sigma^2}$, and $Z = \sqrt{2\pi\sigma^2}$) $$p^*(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$
$X \sim \mathcal{N}(\mu, \sigma^2)$ is mathematically the single most disordered, least-biased, and highest-entropy distribution possible for a given mean and variance continuous system.

###Q. Multiclass Logistic Regression: Automatic Classification of Emails

 $x = (x_1, x_2, x_3)^T$.

According to Jaynes' Principle of Maximum Entropy, because we only have explicit data constraints regarding the **Technical Support** destination ($Y=1$), the probabilities for the other three destinations ($Y=0, 2, 3$) will be distributed uniformly across the remaining probability space:

$$P(Y=0|x) = P(Y=2|x) = P(Y=3|x) = \frac{1 - P(Y=1|x)}{3}$$

Thus, this complex multiclass problem simplifies directly into finding the parameter weights for a binary logistic sigmoid function for $P(Y=1|x)$:

$$P(Y=1|x) = \sigma(w_1 x_1 + w_2 x_2 + w_3 x_3 + b) = \frac{1}{1 + e^{-(w_1 x_1 + w_2 x_2 + w_3 x_3 + b)}}$$





Let $p_s = P(Y=1|x_s)$

*   **Scenario A (Pure Financial):** $x_A = (1,0,0)^T \implies p_A = \sigma(w_1 + b)$
*   **Scenario B (Pure Access):** $x_B = (0,1,0)^T \implies p_B = \sigma(w_2 + b)$
*   **Scenario C (Pure Technical):** $x_C = (0,0,1)^T \implies p_C = \sigma(w_3 + b)$
*   **Scenario D (Access Technical):** $x_D = (0,1,1)^T \implies p_D = \sigma(w_2 + w_3 + b)$
*   **Scenario E (Billing Access):** $x_E = (1,1,0)^T \implies p_E = \sigma(w_1 + w_2 + b)$
*   If an email contains no detected keywords ($x_F = (0,0,0)^T$), its baseline probability is determined solely by the bias: $p_F = \sigma(b)$.*





Using the historical joint expectations provided in your audit logs, we can set up three algebraic equations. The expectation $\mathbb{E}[x_i \cdot I(Y=1)]$ is calculated by summing $P(x) \cdot P(Y=1|x) \cdot x_i$ across all scenarios where feature $x_i = 1$:

#### **Constraint 1: Financial Context**

$$\mathbb{E}[x_1 \cdot I(Y=1)] = P(x_A) \cdot p_A + P(x_E) \cdot p_E = 0.05$$

$$0.40 p_A + 0.05 p_E = 0.05 \implies 8p_A + p_E = 1$$

#### **Constraint 2: Account Access Context**

$$\mathbb{E}[x_2 \cdot I(Y=1)] = P(x_B) \cdot p_B + P(x_D) \cdot p_D + P(x_E) \cdot p_E = 0.15$$

$$0.25 p_B + 0.15 p_D + 0.05 p_E = 0.15 \implies 5p_B + 3p_D + p_E = 3$$

#### **Constraint 3: Bug/Malfunction Context**

$$\mathbb{E}[x_3 \cdot I(Y=1)] = P(x_C) \cdot p_C + P(x_D) \cdot p_D = 0.16$$

$$0.05 p_C + 0.15 p_D = 0.16 \implies p_C + 3p_D = 3.2$$


1.  **From Constraint 3 ($p_C + 3p_D = 3.2$):** Since a probability cannot exceed $1.0$, the maximum value for $p_C$ is $1$. If $p_C = 1$, then $3p_D = 2.2 \implies p_D \approx 0.733$. Therefore, **$p_D$ must be at least $73.3\%$**. If $p_D = 1$, then $p_C = 0.2$. Thus, **$p_C$ must be between $0.20$ and $1.0$**.
2.  **From Constraint 1 ($8p_A + p_E = 1$):** If $p_E = 0$, then $8p_A = 1 \implies p_A = 0.125$. Therefore, **$p_A$ cannot exceed $12.5\%$**. This perfectly aligns with the audit's note that financial keywords heavily suppress routing to Technical support.



```python
import numpy as np
from scipy.optimize import minimize

# Scenario probabilities P(x)
P = {'A': 0.40, 'B': 0.25, 'C': 0.05, 'D': 0.15, 'E': 0.05}

def loss_function(params):
    w1, w2, w3, b = params

    # Define sigmoids for each scenario
    sigma = lambda z: 1 / (1 + np.exp(-z))
    p_A = sigma(w1 + b)
    p_B = sigma(w2 + b)
    p_C = sigma(w3 + b)
    p_D = sigma(w2 + w3 + b)
    p_E = sigma(w1 + w2 + b)

    # Calculate predicted expectations
    exp1 = P['A']*p_A + P['E']*p_E
    exp2 = P['B']*p_B + P['D']*p_D + P['E']*p_E
    exp3 = P['C']*p_C + P['D']*p_D

    # Penalty for violating constraints (squared error)
    loss = (exp1 - 0.05)**2 + (exp2 - 0.15)**2 + (exp3 - 0.16)**2
    return loss

# Optimize parameters
init_guess = [0.0, 0.0, 0.0, 0.0]
result = minimize(loss_function, init_guess, method='BFGS')
w1, w2, w3, b = result.x

print(f"Optimal Weights: w1={w1:.3f}, w2={w2:.3f}, w3={w3:.3f}, bias={b:.3f}")
```
