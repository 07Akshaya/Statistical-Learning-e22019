### 1. Probability Space ($(\Omega,\mathfrak F,P)$)

The sample space is

$$ \Omega={\omega_1,\omega_2,\omega_3,\omega_4}. $$

The $\sigma$-algebra is the power set:

$$ \mathfrak F=\mathfrak P(\Omega), $$

which contains all subsets of ($\Omega$).

The probability measure ($P$) is defined by

$$ P({\omega_1})=0.50,
\qquad
P({\omega_2})=0.10,
\qquad
P({\omega_3})=0.20,
\qquad
P({\omega_4})=0.20. $$

Check:

$$ 0.50+0.10+0.20+0.20=1. $$

Hence,

$$ (\Omega,\mathfrak F,P) $$

is a valid probability space.

---

### 2. Compute ($\sigma(Y)$)

The random variable ($Y$) is

$$ Y(\omega_1)=0,\quad
Y(\omega_2)=0,\quad
Y(\omega_3)=1,\quad
Y(\omega_4)=1. $$

So the partition induced by ($Y$) is

$$ {\omega_1,\omega_2},\qquad
{\omega_3,\omega_4}. $$

Therefore,

$$ \sigma(Y)=
\Big\{
\varnothing,
\Omega,
{\omega_1,\omega_2},
{\omega_3,\omega_4}
\Big\}. $$

---

### 3. Why ($E[X\mid Y]$) must be ($\sigma(Y)$)-measurable

By definition, conditional expectation ($E[X\mid Y]$) must be measurable with respect to ($\sigma(Y)$).

A ($\sigma(Y)$)-measurable random variable can only distinguish between the atoms of ($\sigma(Y)$):

$$ {\omega_1,\omega_2}
\quad\text{and}\quad
{\omega_3,\omega_4}. $$

Therefore ($E[X\mid Y]$) must be constant on each set:

$$ E[X\mid Y](\omega_1) = E[X\mid Y](\omega_2), $$

and

$$ E[X\mid Y](\omega_3) = E[X\mid Y](\omega_4). $$

So let

$$ E[X\mid Y]
=
a\mathbf 1_{{\omega_1,\omega_2}}
+
b\mathbf 1_{{\omega_3,\omega_4}}. $$

---

### 4. Compute ($E[X\mid Y]$)

To find $a$ and $b$, we use the defining property of conditional expectation.

On the set $\{\omega_1,\omega_2\}$,

$$
\int_{\{\omega_1,\omega_2\}} E[X\mid Y]\,dP=\int_{\{\omega_1,\omega_2\}} X\,dP.
$$

The left-hand side is

$$
a\cdot P(\{\omega_1,\omega_2\})=a(0.50+0.10)=0.60a.
$$

The right-hand side is

$$
0\cdot 0.50 + 1\cdot 0.10 = 0.10.
$$

So

$$
0.60a=0.10,
$$

hence

$$
a=\frac{0.10}{0.60}=\frac{1}{6}.
$$

On the set $\{\omega_3,\omega_4\}$,

$$
\int_{\{\omega_3,\omega_4\}} E[X\mid Y]\,dP=\int_{\{\omega_3,\omega_4\}} X\,dP.
$$

The left-hand side is

$$
b\cdot P(\{\omega_3,\omega_4\})=b(0.20+0.20)=0.40b.
$$

The right-hand side is

$$
0\cdot 0.20 + 1\cdot 0.20 = 0.20.
$$

So

$$
0.40b=0.20,
$$

hence

$$
b=\frac{0.20}{0.40}=\frac{1}{2}.
$$

Therefore

$$
E[X\mid Y](\omega)=
\begin{cases}
\dfrac{1}{6}, & \omega\in\{\omega_1,\omega_2\},\\
\dfrac{1}{2}, & \omega\in\{\omega_3,\omega_4\}.
\end{cases}
$$

Equivalently, as a function of $Y$,

$$
E[X\mid Y]=
\frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}=
\begin{cases}
\dfrac{1}{6}, & Y=0,\\
\dfrac{1}{2}, & Y=1.
\end{cases}
$$

Equal.

## For ($A_2$)

0.20. $$

Right side:

$$ \int_{A_2}X,dP
=
0(0.20)+1(0.20)

0.20. $$

Equal.

Hence the defining property holds.

### 8. Mean Square Error ($\mathrm{MSE}$)

The optimal estimator based on ($Y$) is

$$ \hat X=E[X\mid Y]. $$

The $\mathrm{MSE}$ is

$$ \mathrm{MSE}
=
E\left[(X-E[X\mid Y])^2\right]. $$

Using conditional variance:

$$ \mathrm{MSE}
=
E[\mathrm{Var}(X\mid Y)]. $$

---

#### Compute conditional variances

Since ($X$) is Bernoulli,

$$ \mathrm{Var}(X\mid Y=y)
=
p_y(1-p_y). $$

### For ($Y=0$)

$$ p_0=\frac16, $$

$$ \mathrm{Var}(X\mid Y=0)
=
\frac16\cdot\frac56
=
\frac5{36}. $$

### For ($Y=1$)

$$ p_1=\frac12, $$

$$ \mathrm{Var}(X\mid Y=1)
=
\frac12\cdot\frac12
=
\frac14. $$

Now average:

$$ \mathrm{MSE}
=
0.60\cdot\frac5{36}
+
0.40\cdot\frac14. $$

$$ \frac{3}{5}\cdot\frac5{36}
+
\frac25\cdot\frac14. $$

$$ \frac1{12}+\frac1{10}
=
\frac{5+6}{60}
=
\frac{11}{60}. $$

Thus

$$ \boxed{\mathrm{MSE}=\frac{11}{60}\approx0.1833.} $$

---

### 9. MSE using only the global mean

The global estimator is

$$ E[X]=P(X=1)=0.10+0.20=0.30. $$

So estimate every transaction by ($0.30$).

Then

$$ \mathrm{MSE}_{\text{global}}
=
E[(X-0.30)^2]. $$

Since ($X$) is Bernoulli with parameter ($0.30$),

$$ \mathrm{MSE}_{\text{global}}
=
\mathrm{Var}(X)
=
0.30(0.70)
=
0.21. $$

Thus

$$ \boxed{\mathrm{MSE}_{\text{global}}=0.21.} $$

Since

$$ 0.1833 < 0.21, $$

using the risk class improves prediction.

---

#### Relation to the Law of Total Variance

The law states:

$$ \mathrm{Var}(X)
=
E[\mathrm{Var}(X\mid Y)]
+
\mathrm{Var}(E[X\mid Y]). $$

Here,

$$ 0.21
=
\frac{11}{60}
+
\mathrm{Var}(E[X\mid Y]). $$

Hence

$$ \mathrm{Var}(E[X\mid Y])
=
0.21-\frac{11}{60}
=
\frac{19}{2100}
\approx0.0267. $$

This reduction in error is exactly the information gained from knowing ($Y$).

### 10. Information Theory Quantities

---

#### (a) Entropy ($H(X)$)

###Q2) Conditional Density of the Multivariate Gaussian

### 1. Definition

Let

$$
\mathbf Z=
\begin{bmatrix}
\mathbf X\\
\mathbf Y
\end{bmatrix}
\in \mathbb R^{n+m}
$$

be jointly Gaussian with mean

$$
\boldsymbol\mu=
\begin{bmatrix}
\boldsymbol\mu_X\\
\boldsymbol\mu_Y
\end{bmatrix}
$$

and covariance matrix

$$
\boldsymbol\Sigma=
\begin{bmatrix}
\boldsymbol\Sigma_{XX} & \boldsymbol\Sigma_{XY}\\
\boldsymbol\Sigma_{YX} & \boldsymbol\Sigma_{YY}
\end{bmatrix},
$$

where

$$
\boldsymbol\Sigma_{YX}=\boldsymbol\Sigma_{XY}^T.
$$

The joint density is

$$
f_{\mathbf X,\mathbf Y}(\mathbf z)
=
\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol\Sigma|}}
\exp\left(
-\frac12
(\mathbf z-\boldsymbol\mu)^T
\boldsymbol\Sigma^{-1}
(\mathbf z-\boldsymbol\mu)
\right).
$$

### 2. Marginal Distributions

Define

$$
\tilde{\mathbf x}=\mathbf x-\boldsymbol\mu_X,
\qquad
\tilde{\mathbf y}=\mathbf y-\boldsymbol\mu_Y.
$$

Using the Schur complement

$$
\mathbf S_{YY}
=
\boldsymbol\Sigma_{YY}
-\boldsymbol\Sigma_{YX}
\boldsymbol\Sigma_{XX}^{-1}
\boldsymbol\Sigma_{XY},
$$

the quadratic form becomes

$$
Q
=
\tilde{\mathbf x}^T
\boldsymbol\Sigma_{XX}^{-1}
\tilde{\mathbf x}
+
(\tilde{\mathbf y}-A\tilde{\mathbf x})^T
\mathbf S_{YY}^{-1}
(\tilde{\mathbf y}-A\tilde{\mathbf x}),
$$

where

$$
A=\boldsymbol\Sigma_{YX}\boldsymbol\Sigma_{XX}^{-1}.
$$

Hence

$$
f_{\mathbf X,\mathbf Y}(\mathbf x,\mathbf y)
=
C
\exp\left(
-\frac12
\tilde{\mathbf x}^T
\boldsymbol\Sigma_{XX}^{-1}
\tilde{\mathbf x}
\right)
\exp\left(
-\frac12
(\tilde{\mathbf y}-A\tilde{\mathbf x})^T
\mathbf S_{YY}^{-1}
(\tilde{\mathbf y}-A\tilde{\mathbf x})
\right).
$$

Integrating over ($\mathbf y$),

$$
f_{\mathbf X}(\mathbf x)
=
\int_{\mathbb R^m}
f_{\mathbf X,\mathbf Y}(\mathbf x,\mathbf y)\,d\mathbf y.
$$

Using

$$
|\boldsymbol\Sigma|
=
|\boldsymbol\Sigma_{XX}|
|\mathbf S_{YY}|,
$$

we obtain

$$
f_{\mathbf X}(\mathbf x)
=
\frac{1}{\sqrt{(2\pi)^n|\boldsymbol\Sigma_{XX}|}}
\exp\left(
-\frac12
(\mathbf x-\boldsymbol\mu_X)^T
\boldsymbol\Sigma_{XX}^{-1}
(\mathbf x-\boldsymbol\mu_X)
\right).
$$

Thus

$$
\boxed{
\mathbf X\sim\mathcal N(\boldsymbol\mu_X,\boldsymbol\Sigma_{XX})
}
$$

Similarly,

$$
\boxed{
\mathbf Y\sim\mathcal N(\boldsymbol\mu_Y,\boldsymbol\Sigma_{YY})
}
$$

with density

$$
f_{\mathbf Y}(\mathbf y)
=
\frac{1}{\sqrt{(2\pi)^m|\boldsymbol\Sigma_{YY}|}}
\exp\left(
-\frac12
(\mathbf y-\boldsymbol\mu_Y)^T
\boldsymbol\Sigma_{YY}^{-1}
(\mathbf y-\boldsymbol\mu_Y)
\right).
$$

### 3. Conditional Density
To derive the conditional density $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, we use the definition of conditional probability for densities:

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$

Since we have already decomposed the joint density and the marginal density of $\mathbf{Y}$ in the previous steps, this becomes a task of algebraic subtraction in the exponent.

---

From our previous derivation, we factored the joint quadratic form $Q$ using the Schur complement of $\boldsymbol{\Sigma}_{YY}$:
$$Q_{joint} = (\mathbf{y} - \boldsymbol{\mu}_y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_y) + (\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{S}_{XX}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)$$

Where:
* $\mathbf{S}_{XX} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

The marginal density $f_{\mathbf{Y}}(\mathbf{y})$ has the exponent:
$$Q_{marginal} = (\mathbf{y} - \boldsymbol{\mu}_y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_y)$$

When we divide the densities, we subtract the exponents ($e^{A} / e^{B} = e^{A-B}$):
$$Q_{conditional} = Q_{joint} - Q_{marginal} = (\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{S}_{XX}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)$$

---

Now we look at the coefficients.
* **Joint constant:** $\frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}}$
* **Marginal constant:** $\frac{1}{\sqrt{(2\pi)^m |\boldsymbol{\Sigma}_{YY}|}}$

Dividing the joint constant by the marginal constant gives:
$$\frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \cdot \sqrt{(2\pi)^m |\boldsymbol{\Sigma}_{YY}|} = \frac{1}{\sqrt{(2\pi)^n \frac{|\boldsymbol{\Sigma}|}{|\boldsymbol{\Sigma}_{YY}|}}}$$

Using the determinant identity $|\boldsymbol{\Sigma}| = |\boldsymbol{\Sigma}_{YY}| \cdot |\mathbf{S}_{XX}|$, the term $\frac{|\boldsymbol{\Sigma}|}{|\boldsymbol{\Sigma}_{YY}|}$ simplifies exactly to $|\mathbf{S}_{XX}|$.

---

Combining the simplified constant and the remaining quadratic form, we get:

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{S}_{XX}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right)$$

This is the density of a multivariate normal distribution:
$$(\mathbf{X} \mid \mathbf{Y} = \mathbf{y}) \sim \mathscr{N}(\boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y), \quad \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}).$$

### 4. Conditional Expectation

Since

$$
(\mathbf X|\mathbf Y=\mathbf y)
\sim
\mathcal N(\boldsymbol\mu^*,\mathbf S_{XX}),
$$

its mean is

$$
E[\mathbf X|\mathbf Y=\mathbf y]
=
\boldsymbol\mu_X
+
\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
(\mathbf y-\boldsymbol\mu_Y).
$$

Hence

$$
\boxed{
E[\mathbf X|\mathbf Y]
=
\boldsymbol\mu_X
+
\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
(\mathbf Y-\boldsymbol\mu_Y)
}
$$

### 5. Conditional Variance

The conditional covariance matrix is

$$
\boxed{
\operatorname{Var}(\mathbf X|\mathbf Y)
=
\boldsymbol\Sigma_{XX}
-\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}
}
$$

and therefore

$$
E[\operatorname{Var}(\mathbf X|\mathbf Y)]
=
\boldsymbol\Sigma_{XX}
-\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}.
$$

The estimation error is

$$
\boxed{
E\left[
(\mathbf X-E[\mathbf X|\mathbf Y])
(\mathbf X-E[\mathbf X|\mathbf Y])^T
\right]
=
\boldsymbol\Sigma_{XX}
-\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}
}
$$

Using the law of total variance,

$$
\operatorname{Var}(\mathbf X)
=
E[\operatorname{Var}(\mathbf X|\mathbf Y)]
+
\operatorname{Var}(E[\mathbf X|\mathbf Y]),
$$

we obtain

$$
\boxed{
\operatorname{Var}(E[\mathbf X|\mathbf Y])
=
\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}
}
$$

### 6. Differential Entropy and Mutual Information

#### 6.1 Differential Entropy of a Multivariate Gaussian

For

$$
\mathbf Z\sim\mathcal N(\boldsymbol\mu,\boldsymbol\Sigma),
$$

the differential entropy is

$$
h(\mathbf Z)
=
-\mathbb E[\ln f_{\mathbf Z}(\mathbf Z)].
$$

Using the Gaussian density,

$$
\boxed{
h(\mathbf Z)
=
\frac12
\ln\left(
(2\pi e)^{n+m}
|\boldsymbol\Sigma|
\right)
}
$$

#### 6.2 Conditional Entropy

Since

$$
(\mathbf X|\mathbf Y=\mathbf y)
\sim
\mathcal N(\boldsymbol\mu^*,\mathbf S_{XX}),
$$

the conditional entropy is

$$
\boxed{
h(\mathbf X|\mathbf Y)
=
\frac12
\ln\left(
(2\pi e)^n
|\mathbf S_{XX}|
\right)
}
$$

where

$$
\mathbf S_{XX}
=
\boldsymbol\Sigma_{XX}
-\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}.
$$

#### 6.3 Entropy of the Conditional Expectation

Since

$$
\operatorname{Var}(E[\mathbf X|\mathbf Y])
=
\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX},
$$

the entropy is

$$
\boxed{
h(E[\mathbf X|\mathbf Y])
=
\frac12
\ln\left(
(2\pi e)^n
\left|
\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}
\right|
\right)
}
$$

#### 6.4 Mutual Information

Mutual information is

$$
I(\mathbf X;\mathbf Y)
=
h(\mathbf X)-h(\mathbf X|\mathbf Y).
$$

Substituting the entropy formulas,

$$
I(\mathbf X;\mathbf Y)
=
\frac12
\ln
\left(
\frac{
|\boldsymbol\Sigma_{XX}|
}{
|\mathbf S_{XX}|
}
\right).
$$

Hence,

$$
\boxed{
I(\mathbf X;\mathbf Y)
=
\frac12
\ln
\left(
\frac{
|\boldsymbol\Sigma_{XX}|
}{
|\boldsymbol\Sigma_{XX}
-\boldsymbol\Sigma_{XY}
\boldsymbol\Sigma_{YY}^{-1}
\boldsymbol\Sigma_{YX}|
}
\right)
}
$$
