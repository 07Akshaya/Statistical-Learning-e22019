### 1. Probability Space ($(\Omega,\mathfrak F,P)$)

The sample space is

$$ \Omega={\omega_1,\omega_2,\omega_3,\omega_4}. $$

The $\sigma$-algebra is the power set:

$$ \mathfrak F=\mathfrak P(\Omega), $$

which contains all subsets of ($\Omega$).

The probability measure ($P$) is defined by

$$ P({\omega_1})=0.50,
\qquad
P({\omega_2})=0.10,
\qquad
P({\omega_3})=0.20,
\qquad
P({\omega_4})=0.20. $$

Check:

$$ 0.50+0.10+0.20+0.20=1. $$

Hence,

$$ (\Omega,\mathfrak F,P) $$

is a valid probability space.

### 2. Compute ($\sigma(Y)$)

The random variable ($Y$) is

$$ Y(\omega_1)=0,\quad
Y(\omega_2)=0,\quad
Y(\omega_3)=1,\quad
Y(\omega_4)=1. $$

So the partition induced by ($Y$) is

$$ {\omega_1,\omega_2},\qquad
{\omega_3,\omega_4}. $$

Therefore,

$$ \sigma(Y)=
\Big\{
\varnothing,
\Omega,
{\omega_1,\omega_2},
{\omega_3,\omega_4}
\Big\}. $$

---

### 3. Why ($E[X\mid Y]$) must be ($\sigma(Y)$)-measurable

By definition, conditional expectation ($E[X\mid Y]$) must be measurable with respect to ($\sigma(Y)$).

A ($\sigma(Y)$)-measurable random variable can only distinguish between the atoms of ($\sigma(Y)$):

$$ {\omega_1,\omega_2}
\quad\text{and}\quad
{\omega_3,\omega_4}. $$

Therefore ($E[X\mid Y]$) must be constant on each set:

$$ E[X\mid Y](\omega_1) = E[X\mid Y](\omega_2), $$

and

$$ E[X\mid Y](\omega_3) = E[X\mid Y](\omega_4). $$

So let

$$ E[X\mid Y]
=
a\mathbf 1_{{\omega_1,\omega_2}}
+
b\mathbf 1_{{\omega_3,\omega_4}}. $$

---

### 4. Compute ($E[X\mid Y]$)

To find $a$ and $b$, we use the defining property of conditional expectation.

On the set $\{\omega_1,\omega_2\}$,

$$
\int_{\{\omega_1,\omega_2\}} E[X\mid Y]\,dP=\int_{\{\omega_1,\omega_2\}} X\,dP.
$$

The left-hand side is

$$
a\cdot P(\{\omega_1,\omega_2\})=a(0.50+0.10)=0.60a.
$$

The right-hand side is

$$
0\cdot 0.50 + 1\cdot 0.10 = 0.10.
$$

So

$$
0.60a=0.10,
$$

hence

$$
a=\frac{0.10}{0.60}=\frac{1}{6}.
$$

On the set $\{\omega_3,\omega_4\}$,

$$
\int_{\{\omega_3,\omega_4\}} E[X\mid Y]\,dP=\int_{\{\omega_3,\omega_4\}} X\,dP.
$$

The left-hand side is

$$
b\cdot P(\{\omega_3,\omega_4\})=b(0.20+0.20)=0.40b.
$$

The right-hand side is

$$
0\cdot 0.20 + 1\cdot 0.20 = 0.20.
$$

So

$$
0.40b=0.20,
$$

hence

$$
b=\frac{0.20}{0.40}=\frac{1}{2}.
$$

Therefore

$$
E[X\mid Y](\omega)=
\begin{cases}
\dfrac{1}{6}, & \omega\in\{\omega_1,\omega_2\},\\
\dfrac{1}{2}, & \omega\in\{\omega_3,\omega_4\}.
\end{cases}
$$

Equivalently, as a function of $Y$,

$$
E[X\mid Y]=
\frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}=
\begin{cases}
\dfrac{1}{6}, & Y=0,\\
\dfrac{1}{2}, & Y=1.
\end{cases}
$$

### 5. Verify the defining property

We must check:

$$ \int_A E[X\mid Y],dP
=
\int_A X,dP $$

for all ($A\in\sigma(Y)$).

The nontrivial sets are:

$$ A_1={\omega_1,\omega_2},
\qquad
A_2={\omega_3,\omega_4}. $$

---

## For ($A_1$)

Left side:

$$ \int_{A_1}E[X\mid Y],dP
=
\frac16(0.50+0.10)
=
\frac16(0.60)
=
0.10. $$

Right side:

$$ \int_{A_1}X,dP
=
0(0.50)+1(0.10)
=
0.10. $$

Equal.

---

## For ($A_2$)

Left side:

$$ \int_{A_2}E[X\mid Y],dP
=
\frac12(0.20+0.20)
=
0.20. $$

Right side:

$$ \int_{A_2}X,dP
=
0(0.20)+1(0.20)
=
0.20. $$

Equal.

Hence the defining property holds.

### 8. Mean Square Error ($\mathrm{MSE}$)

The optimal estimator based on ($Y$) is

$$ \hat X=E[X\mid Y]. $$

The $\mathrm{MSE}$ is

$$ \mathrm{MSE}
=
E\left[(X-E[X\mid Y])^2\right]. $$

Using conditional variance:

$$ \mathrm{MSE}
=
E[\mathrm{Var}(X\mid Y)]. $$

---

#### Compute conditional variances

Since ($X$) is Bernoulli,

$$ \mathrm{Var}(X\mid Y=y)
=
p_y(1-p_y). $$

### For ($Y=0$)

$$ p_0=\frac16, $$

$$ \mathrm{Var}(X\mid Y=0)
=
\frac16\cdot\frac56
=
\frac5{36}. $$

### For ($Y=1$)

$$ p_1=\frac12, $$

$$ \mathrm{Var}(X\mid Y=1)
=
\frac12\cdot\frac12
=
\frac14. $$

Now average:

$$ \mathrm{MSE}
=
0.60\cdot\frac5{36}
+
0.40\cdot\frac14. $$

$$ \frac{3}{5}\cdot\frac5{36}
+
\frac25\cdot\frac14. $$

$$ \frac1{12}+\frac1{10}
=
\frac{5+6}{60}
=
\frac{11}{60}. $$

Thus

$$ \boxed{\mathrm{MSE}=\frac{11}{60}\approx0.1833.} $$

---

### 9. MSE using only the global mean

The global estimator is

$$ E[X]=P(X=1)=0.10+0.20=0.30. $$

So estimate every transaction by ($0.30$).

Then

$$ \mathrm{MSE}_{\text{global}}
=
E[(X-0.30)^2]. $$

Since ($X$) is Bernoulli with parameter ($0.30$),

$$ \mathrm{MSE}_{\text{global}}
=
\mathrm{Var}(X)
=
0.30(0.70)
=
0.21. $$

Thus

$$ \boxed{\mathrm{MSE}_{\text{global}}=0.21.} $$

Since

$$ 0.1833 < 0.21, $$

using the risk class improves prediction.

---

#### Relation to the Law of Total Variance

The law states:

$$ \mathrm{Var}(X)
=
E[\mathrm{Var}(X\mid Y)]
+
\mathrm{Var}(E[X\mid Y]). $$

Here,

$$ 0.21
=
\frac{11}{60}
+
\mathrm{Var}(E[X\mid Y]). $$

Hence

$$ \mathrm{Var}(E[X\mid Y])
=
0.21-\frac{11}{60}
=
\frac{19}{2100}
\approx0.0267. $$

This reduction in error is exactly the information gained from knowing ($Y$).

### 10. Information Theory Quantities

---

#### (a) Entropy ($H(X)$)

($X\sim \text{Bernoulli}(0.30)$).

$$ H(X)
=
-\sum_x p(x)\log_2 p(x). $$

So

$$ H(X)
=
-0.7\log_2(0.7)-0.3\log_2(0.3). $$

$$ \boxed{H(X)\approx0.881\text{ bits}.} $$

---

#### (b) Conditional Entropy ($H(X\mid Y)$)

$$ H(X\mid Y)
=
\sum_y P(Y=y)H(X\mid Y=y). $$

---

### For ($Y=0$)

$$ H(X\mid Y=0)
=
-\frac56\log_2\frac56
-\frac16\log_2\frac16. $$

$$ \approx0.650\text{ bits}. $$

---

##### For ($Y=1$)

$$ H(X\mid Y=1)
=
1\text{ bit}. $$

Therefore

$$ H(X\mid Y)
=
0.60(0.650)+0.40(1). $$

$$ \boxed{H(X\mid Y)\approx0.790\text{ bits}.} $$

---

#### (c) Mutual Information ($I(X;Y)$)

$$ I(X;Y)
=
H(X)-H(X\mid Y). $$

Thus

$$ I(X;Y)
=
0.881-0.790. $$

$$ \boxed{I(X;Y)\approx0.091\text{ bits}.} $$

---

#### (d) Percentage of uncertainty removed

$$ \frac{I(X;Y)}{H(X)}\times100%. $$

$$ \frac{0.091}{0.881}\times100%. $$

$$ \boxed{\approx10.3%.} $$

So knowing the customer’s risk class removes about **10% of the uncertainty** about whether a transaction is fraudulent.
