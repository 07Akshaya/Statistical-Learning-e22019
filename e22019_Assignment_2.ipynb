## Q3) Transformer Failures

### 1. Probability Space

A suitable probability space is

$$
\Omega = \{0,1,2,3\}
$$

$$
\mathfrak F = \mathfrak P(\Omega)
$$

where $\mathfrak P(\Omega)$ is the power set of $\Omega$.

Define the probability measure $P$ by

$$
P(\{0\}) = 0.50
$$

$$
P(\{1\}) = 0.30
$$

$$
P(\{2\}) = 0.15
$$

$$
P(\{3\}) = 0.05
$$

For any event $A \subseteq \Omega$,

$$
P(A) = \sum_{\omega \in A} P(\{\omega\})
$$



### 2. Definition of the Random Variable

Define the random variable

$$
X : \Omega \to \mathbb{R}
$$

by

$$
X(\omega) = \omega, \qquad \omega \in \Omega
$$

Here, $X$ represents the number of emergency transformer failures reported in one day.



### 3. Probability Mass Function (PMF)

The probability mass function of $X$ is

$$
p_X(x) = P(X=x)
$$

Thus,

$$
p_X(x)=
\begin{cases}
0.50, & x=0 \\
0.30, & x=1 \\
0.15, & x=2 \\
0.05, & x=3 \\
0, & \text{otherwise}
\end{cases}
$$



### 4. Expectation $$E[X]$$

The expectation is

$$
E[X] = \sum_x x\,p_X(x)
$$

$$
= 0(0.50)+1(0.30)+2(0.15)+3(0.05)
$$

$$
= 0 + 0.30 + 0.30 + 0.15
$$

$$
E[X] = 0.75
$$

So,

$$
\boxed{E[X]=0.75}
$$



### 5. Variance $$\mathrm{Var}(X)$$

First compute $$E[X^2]$$:

$$
E[X^2] = \sum_x x^2 p_X(x)
$$

$$
=0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)
$$

$$
=0+0.30+0.60+0.45
$$

$$
E[X^2]=1.35
$$

Now,

$$
\mathrm{Var}(X)=E[X^2]-(E[X])^2
$$

$$
=1.35-(0.75)^2
$$

$$
=1.35-0.5625
$$

$$
\boxed{\mathrm{Var}(X)=0.7875}
$$



### 6. Required Probabilities

#### (a) Probability that at least one failure occurs

$$
P(X \ge 1)=P(X=1)+P(X=2)+P(X=3)
$$

$$
=0.30+0.15+0.05
$$

$$
\boxed{P(X \ge 1)=0.50}
$$



#### (b) Probability that at least two failures occur

$$
P(X \ge 2)=P(X=2)+P(X=3)
$$

$$
=0.15+0.05
$$

$$
\boxed{P(X \ge 2)=0.20}
$$



### 7. Practical Meaning

The expectation

$$
E[X]=0.75
$$

means that the average number of emergency transformer failures per day is $$0.75$$.

The variance

$$
\mathrm{Var}(X)=0.7875
$$

measures the day-to-day variability in the number of failures. A larger variance would indicate greater fluctuation in the daily number of transformer failures.

### Plot of the triangular distribution

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang

# Parameters
a = 4          # minimum
b = 10         # maximum
c_mode = 6     # mode (most likely value)

# Scipy triangular distribution parameters
loc = a
scale = b - a
c_scipy = (c_mode - a) / (b - a)

# Generate x values only within support
x = np.linspace(a, b, 500)

# PDF values
y = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)

# Create figure
fig = go.Figure()

# Distribution curve
fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    fill='tozeroy',
    name='Triangular PDF',
    line=dict(color='lavender', width=3)
))

# Add vertical lines for a, c, b
fig.add_vline(x=a, line_dash="dash", line_color="green")
fig.add_vline(x=c_mode, line_dash="dash", line_color="red")
fig.add_vline(x=b, line_dash="dash", line_color="orange")

# Add annotations
fig.add_annotation(x=a, y=max(y)*0.2,
                   text="a = 4<br>Minimum",
                   showarrow=True)

fig.add_annotation(x=c_mode, y=max(y),
                   text="c = 6<br>Mode",
                   showarrow=True)

fig.add_annotation(x=b, y=max(y)*0.2,
                   text="b = 10<br>Maximum",
                   showarrow=True)

# Layout
fig.update_layout(
    title='Triangular Distribution of Installation Time',
    xaxis_title='Installation Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white',
    width=900,
    height=500
)

# Show plot
fig.show()
```

##Q6) Expected Project Completion Time Using PERT



### 1. Compute the values of the shape parameters (α) and (β)

The given parameters are

$$
a=8, \qquad b=20, \qquad c=11.
$$

Using the PERT formulas,

$$
\alpha = 1 + 4\left(\frac{c-a}{b-a}\right)
$$

we obtain

$$
\alpha
=1+4\left(\frac{11-8}{20-8}\right)
=1+4\left(\frac{3}{12}\right)
=1+1
=2.
$$

Similarly,

$$
\beta = 1 + 4\left(\frac{b-c}{b-a}\right)
$$

gives

$$
\beta
=1+4\left(\frac{20-11}{20-8}\right)
=1+4\left(\frac{9}{12}\right)
=1+3
=4.
$$

Therefore,

$$
\boxed{\alpha=2,\qquad \beta=4.}
$$


### 2. Specify the support of (X)

Since (X) represents the number of days required to complete the task, the possible values lie between the optimistic and pessimistic times.

Thus, the support is

$$
\boxed{8 \le X \le 20.}
$$


### 3. Write down the probability density function ($f_X(x)$)

The PERT density is

$$
f_X(x)=
\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}
{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}},
\qquad a\le x\le b.
$$

Substituting ($a=8$), ($b=20$), ($\alpha=2$), and ($\beta=4$),

$$
f_X(x)
=
\frac{(x-8)^{1}(20-x)^3}
{B(2,4)12^5},
\qquad 8\le x\le20.
$$

Now compute the Beta function:

$$
B(2,4)
=
\frac{\Gamma(2)\Gamma(4)}{\Gamma(6)}
=
\frac{1!3!}{5!}
=
\frac{6}{120}
=
\frac{1}{20}.
$$

Hence,

$$
f_X(x)
=
\frac{20(x-8)(20-x)^3}{12^5},
\qquad 8\le x\le20.
$$

Also,

$$
f_X(x)=0
\qquad \text{otherwise}.
$$

Therefore,

$$
\boxed{
f_X(x)=
\begin{cases}
\dfrac{20(x-8)(20-x)^3}{12^5}, & 8\le x\le20,
0, & \text{otherwise}.
\end{cases}}
$$


### 4. State the probability space (($\Omega$,$\mathfrak F$,P))

A suitable probability space for this continuous model is

$$
\Omega=[8,20],
$$

$$
\mathfrak F=\mathfrak B([8,20]),
$$

where ($\mathfrak B([8,20])$) denotes the Borel ($\sigma$)-algebra on ([8,20]).

Define the probability measure (P) by

$$
P(A)=\int_A f_X(x)\,dx,
\qquad A\in\mathfrak F.
$$

Thus,

$$
\boxed{
(\Omega,\mathfrak F,P)
=
\left(
[8,20],
\mathfrak B([8,20]),
P
\right).
}
$$


### 5. Define (X) as a random variable

Define

$$
X:\Omega\to\mathbb R
$$

by

$$
X(\omega)=\omega,
\qquad \omega\in[8,20].
$$

Thus (X) records the task completion time in days.


### 6. Show that the expectation is

The expectation is defined as:
$$E[X] = \int_{a}^{b} x f(x) \, dx = \int_{a}^{b} x \frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}} \, dx$$

To simplify the integral, we transform the variable $x$ into a standard Beta variable $t$ on the interval $[0, 1]$.
Let:
$$t = \frac{x-a}{b-a} \implies x = a + (b-a)t \quad \text{and} \quad dx = (b-a)dt$$
Substituting the boundaries:
* When $x = a$, $t = 0$.
* When $x = b$, $t = 1$.

Now, substitute $x$, $(x-a)$, and $(b-x)$ in the integral:
* $(x-a) = (b-a)t$
* $(b-x) = b - (a + (b-a)t) = (b-a)(1-t)$

The integral becomes:
$$E[X] = \int_{0}^{1} \big( a + (b-a)t \big) \frac{[(b-a)t]^{\alpha-1}[(b-a)(1-t)]^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}} (b-a) \, dt$$

Pulling out the constant terms $(b-a)$:
$$E[X] = \frac{(b-a)^{\alpha-1+\beta-1+1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}} \int_{0}^{1} \big( a + (b-a)t \big) t^{\alpha-1}(1-t)^{\beta-1} \, dt$$
The powers of $(b-a)$ cancel out exactly:
$$E[X] = \frac{1}{B(\alpha,\beta)} \int_{0}^{1} \big( a t^{\alpha-1}(1-t)^{\beta-1} + (b-a)t^{\alpha}(1-t)^{\beta-1} \big) \, dt$$

Using the definition of the Beta function $B(\alpha, \beta) = \int_0^1 t^{\alpha-1}(1-t)^{\beta-1} \, dt$:
$$E[X] = \frac{a B(\alpha, \beta) + (b-a) B(\alpha+1, \beta)}{B(\alpha, \beta)} = a + (b-a) \frac{B(\alpha+1, \beta)}{B(\alpha, \beta)}$$

Using the identity $B(x, y) = \frac{\Gamma(x)\Gamma(y)}{\Gamma(x+y)}$:
$$\frac{B(\alpha+1, \beta)}{B(\alpha, \beta)} = \frac{\Gamma(\alpha+1)\Gamma(\beta)}{\Gamma(\alpha+\beta+1)} \cdot \frac{\Gamma(\alpha+\beta)}{\Gamma(\alpha)\Gamma(\beta)} = \frac{\alpha \Gamma(\alpha)}{\Gamma(\alpha)} \cdot \frac{\Gamma(\alpha+\beta)}{(\alpha+\beta)\Gamma(\alpha+\beta)} = \frac{\alpha}{\alpha+\beta}$$
Thus:
$$E[X] = a + (b-a) \frac{\alpha}{\alpha+\beta}$$

First, calculate $\alpha + \beta$:
$$\alpha + \beta = \left( 1 + 4\frac{c-a}{b-a} \right) + \left( 1 + 4\frac{b-c}{b-a} \right) = 2 + 4 \left( \frac{c-a+b-c}{b-a} \right) = 2 + 4(1) = 6$$

Now substitute $\alpha$ and $(\alpha+\beta)$ into the expectation formula:
$$E[X] = a + (b-a) \frac{1 + 4\frac{c-a}{b-a}}{6}$$
$$E[X] = a + \frac{b-a}{6} + \frac{4(c-a)}{6}$$
$$E[X] = \frac{6a + b - a + 4c - 4a}{6}$$
$$E[X] = \frac{a + 4c + b}{6}$$




Substituting
$$
E[X]=\frac{a+4c+b}{6}
=\frac{8+4(11)+20}{6}
=\frac{8+44+20}{6}
=\frac{72}{6}
=12.
$$

So

$$
E[X]=12.
$$

### 7. Compute the variance

Using the approximation

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7},
$$

substitute ($E[X]=12$), ($a=8$), and ($b=20$):

$$
\mathrm{Var}(X)
=
\frac{(12-8)(20-12)}{7}.
$$

Thus,

$$
\mathrm{Var}(X)
=
\frac{4\times8}{7}
=
\frac{32}{7}.
$$

Therefore,

$$
\boxed{
\mathrm{Var}(X)=\frac{32}{7}\approx4.57.
}
$$


### 8. Compute the approximate standard deviation

The standard deviation is

$$
\sigma=\sqrt{\mathrm{Var}(X)}.
$$

Hence,

$$
\sigma
=
\sqrt{\frac{32}{7}}
\approx2.14.
$$

Therefore,

$$
\boxed{\sigma\approx2.14\text{ days}.}
$$

Also, using the common PERT approximation,

$$
\sigma\approx\frac{b-a}{6}
=
\frac{20-8}{6}
=2,
$$

which is close to the calculated value.


### 9. Why is the PERT distribution more appropriate than the Triangular distribution?

For comparison:

| Quantity           |    Triangular Distribution |          PERT Distribution |
| :----------------- | -------------------------: | -------------------------: |
| Mean               |                       13 |                       12 |
| Variance           | $\dfrac{49}{6}\approx8.17$ | $\dfrac{32}{7}\approx4.57$ |
| Standard deviation |              $\approx2.86$ |              $\approx2.14$ |
| Shape              |           sharp peak  linear |         Smooth mode weighted more heavily |

The PERT distribution is often preferred because:

* it gives **more weight to the most likely value** (c),
* it produces a **smoother and more realistic curve**,
* it avoids the sharp corners of the Triangular distribution,
* and it typically gives a **less extreme variance estimate**.

In project management, task durations usually cluster around the most likely value rather than changing linearly, so the PERT model often represents uncertainty more realistically.


### 10. Practical interpretation in project planning

In this context:

* $\boxed{a=8}$ days is the **optimistic time**, meaning the shortest realistic completion time if everything goes well.

* $\boxed{b=20}$ days is the **pessimistic time**, meaning the longest realistic completion time under difficult conditions.

* $\boxed{c=11}$ days is the **most likely time**, representing the duration expected under normal working conditions.

* $\boxed{E[X]=12}$ days is the **expected project duration**, which managers may use for scheduling and planning.

* $\boxed{\sigma\approx2.14}$ days measures the **uncertainty or variability** in completion time. A larger ($\sigma$) would indicate greater scheduling risk.

These quantities help project managers estimate deadlines, allocate resources, and include contingency buffers when planning the project schedule.

### Plot of the triangle and PERT distributions

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang, beta

# Parameters
a = 8
b = 20
c_mode = 11

# -----------------------------------
# Triangular Distribution Parameters
# -----------------------------------

c_scipy = (c_mode - a) / (b - a)

loc = a
scale = b - a

# -----------------------------------
# PERT Distribution Parameters
# -----------------------------------

alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

print("PERT Parameters")
print("alpha =", alpha_pert)
print("beta  =", beta_pert)

# -----------------------------------
# Generate x values
# Start from 0 to check the support
# -----------------------------------

x = np.linspace(0, 25, 1000)

# -----------------------------------
# Probability Density Functions
# -----------------------------------

y_triang = triang.pdf(
    x,
    c=c_scipy,
    loc=loc,
    scale=scale
)

y_pert = beta.pdf(
    x,
    alpha_pert,
    beta_pert,
    loc=a,
    scale=b-a
)

# -----------------------------------
# Create Plot
# -----------------------------------

fig = go.Figure()

# Triangular distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_triang,
    mode='lines',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=3, dash='dash')
))

# PERT distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    name='PERT Distribution',
    line=dict(color='firebrick', width=3)
))

# -----------------------------------
# Add important points
# -----------------------------------

fig.add_vline(
    x=a,
    line_dash="dot",
    line_color="green",
    annotation_text="a = 8"
)

fig.add_vline(
    x=c_mode,
    line_dash="dot",
    line_color="orange",
    annotation_text="c = 11"
)

fig.add_vline(
    x=b,
    line_dash="dot",
    line_color="purple",
    annotation_text="b = 20"
)

# -----------------------------------
# Layout
# -----------------------------------

fig.update_layout(
    title='PERT vs Triangular Distribution',
    xaxis_title='Task Completion Time (days)',
    yaxis_title='Probability Density',
    template='plotly_white',
    width=950,
    height=550
)

# -----------------------------------
# Show Plot
# -----------------------------------

fig.show()
```
