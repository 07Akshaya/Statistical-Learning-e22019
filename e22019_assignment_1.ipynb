##Q1) The Probability of Shooting at a Target

The disk is  
$$
\Omega=\{(x,y)\in \mathbb{R}^2 : x^2+y^2\le 100\}
$$

Let $A$ be the event that the point lies within 1 unit of the boundary.  
This means the distance from the center is between 9 and 10.

$$
A=\{(x,y)\in \Omega : 9 \le \sqrt{x^2+y^2} \le 10\}
$$

So, $A$ is the outer ring.



### 1. Uniform Distribution with Respect to Area

In this model, equal **areas** have equal probability.

- Total area of the disk:
$$
\pi (10)^2 = 100\pi
$$

- Area of event $A$:
$$
\pi (10)^2 - \pi (9)^2 = 100\pi - 81\pi = 19\pi
$$

- Probability:
$$
P(A) = \frac{19\pi}{100\pi} = \frac{19}{100}
$$

$$
\boxed{P(A) = \frac{19}{100}}
$$



### 2. Uniform Distribution of the Radius

Here:
- Radius $r \sim \text{Uniform}[0,10]$
- Angle $\theta \sim \text{Uniform}[0,2\pi)$

We only need to consider $r$.

- Total length of $r$:  
$$
10 - 0 = 10
$$

- Favorable length ($r \in [9,10]$):  
$$
10 - 9 = 1
$$

- Probability:
$$
P(A) = \frac{Favorable Length}{Total Length}=\frac{1}{10}
$$

$$
\boxed{P(A) = \frac{1}{10}}
$$



### 3. Why the Answers Differ

The results are different because “random” is defined differently.

- In the **area model**, probability depends on area.  
  The outer ring has more area → higher probability.

- In the **radius model**, all distances are equally likely.  
  So only interval length matters.

$$
\frac{19}{100} \neq \frac{1}{10}
$$

## Q2) Bertrand's Paradox
### Preliminary Geometric Fact

Consider the unit circle

$$
x^2 + y^2 = 1.
$$

The side length of an inscribed equilateral triangle in this circle is

$$
\sqrt{3}.
$$

We show that a chord is longer than $\sqrt{3}$ **if and only if** its perpendicular distance from the center is less than $\frac{1}{2}$.




### Proof

Let a chord be at a perpendicular distance $r \in [0,1]$ from the center.

From geometry, half of the chord length forms a right triangle with:
- hypotenuse = $1$ (radius),
- one leg = $r$.

Hence, the chord length is

$$
L(r) = 2\sqrt{1 - r^2}.
$$

We require

$$
L(r) > \sqrt{3}.
$$

So,

$$
2\sqrt{1 - r^2} > \sqrt{3}.
$$


$$
\sqrt{1 - r^2} > \frac{\sqrt{3}}{2}.
$$


$$
1 - r^2 > \frac{3}{4}.
$$

$$
r^2 < \frac{1}{4}.
$$


$$
r < \frac{1}{2}.
$$


The required event can be written as

$$
A = \left\{ \text{distance of the chord from the center} < \frac{1}{2} \right\}.
$$



### Model 1: Uniform Endpoints (Angle Model)

Choose one endpoint uniformly and the other uniformly.

Let $\theta \in [0,\pi]$ be the central angle.

$$
L(\theta) = 2\sin(\theta/2)
$$

Condition:

$$
2\sin(\theta/2) > \sqrt{3}
\;\Longleftrightarrow\;
\theta > \frac{2\pi}{3}
$$

Probability:

$$
P = \frac{\pi - 2\pi/3}{\pi} = \frac{1}{3}
$$

 **Result:**

$$
P = \frac{1}{3}
$$



### Model 2: Uniform Distance from Center

Choose distance $r \in [0,1]$ uniformly.

Condition:

$$
r < \frac{1}{2}
$$

Probability:

$$
P = \frac{1}{2}
$$

**Result:**

$$
P = \frac{1}{2}
$$



### Model 3: Uniform Midpoint in Disk

Choose midpoint uniformly in the unit disk:

$$
x^2 + y^2 \leq 1
$$

Condition:

$$
x^2 + y^2 < \frac{1}{4}
$$

Probability:

$$
P = \frac{\pi (1/2)^2}{\pi} = \frac{1}{4}
$$

 **Result:**

$$
P = \frac{1}{4}
$$


### Final Conclusion

| Model | Uniform Choice | Probability |
|------|--------------|------------|
| 1 | Angle | $1/3$ |
| 2 | Distance | $1/2$ |
| 3 | Midpoint | $1/4$ |



The phrase **"random chord" is ambiguous**.

Different reasonable models give different answers:

$$
\frac{1}{3}, \quad \frac{1}{2}, \quad \frac{1}{4}
$$



##Q8) Probability of Being a Girl?


### Sample Space

The sample space is

$$
\Omega = \{BB, BG, GB, GG\},
$$

where:
- the first letter represents the older child,
- the second letter represents the younger child.

Each outcome has probability

$$
\frac{1}{4}.
$$


The condition “at least one child is a girl” eliminates only the outcome $BB$.

Thus, the reduced sample space becomes

$$
\{GG, BG, GB\}.
$$

Each of these outcomes is equally likely.


Out of these three outcomes, only one corresponds to both children being girls ($GG$). Therefore,

$$
P(\text{both girls} \mid \text{at least one girl}) = \frac{1}{3}
$$


$$
\text{Final Answer} = \frac{1}{3}
$$


### Why This Result Is Interesting

Many people initially guess $\frac{1}{2}$, but that corresponds to a *different* situation.

For example, if we are told:

> “The older child is a girl,”

then the possible outcomes reduce to

$$
\{GG, GB\}.
$$

In this case,

$$
P(\text{both girls} \mid \text{older child is a girl}) = \frac{1}{2}.
$$


## Q9) Information Available in a Discrete Random Variable



### 1. Probability Space

The sample space is

$$
\Omega = \{G_1, G_2, D_1, D_2\},
$$

and the σ-algebra is

$$
\mathfrak{F} = \mathfrak{P}(\Omega).
$$

The probability measure is defined as

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20.
$$

Since

$$
0.50 + 0.20 + 0.10 + 0.20 = 1,
$$

this defines a valid probability space.


### 2. σ-algebra Generated by $X$

The random variable $X$ classifies items as good or defective:

$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
$$

Thus,

$$
X^{-1}(\{0\}) = \{G_1, G_2\}, \quad X^{-1}(\{1\}) = \{D_1, D_2\}.
$$

Hence,

$$
\sigma(X) = \{\varnothing, \Omega, \{G_1,G_2\}, \{D_1,D_2\}\}.
$$



### 3. σ-algebra Generated by $Y$

The random variable $Y$ distinguishes all outcomes:

$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
$$

Since each outcome has a unique value,

$$
\sigma(Y) = \mathfrak{P}(\Omega).
$$

Thus, $Y$ generates the full power set.



### 4. Relation Between $\sigma(X)$ and $\sigma(Y)$

Since $\sigma(Y)$ contains all subsets of $\Omega$,

$$
\sigma(X) \subseteq \sigma(Y).
$$

This shows that $Y$ carries strictly more information than $X$, because it identifies both quality and machine, whereas $X$ only identifies quality.



### 5. Marginal Distribution of $X$

$$
P(X=0) = P(G_1) + P(G_2) = 0.50 + 0.20 = 0.70,
$$

$$
P(X=1) = P(D_1) + P(D_2) = 0.10 + 0.20 = 0.30.
$$



### 6. Marginal Distribution of $Y$

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20.
$$


### 7. Joint Probabilities

$$
P(X=0,Y=1)=0.50,\quad P(X=0,Y=2)=0.20,
$$

$$
P(X=1,Y=3)=0.10,\quad P(X=1,Y=4)=0.20.
$$

All other combinations have probability zero.



### 8. Conditional Probabilities

Since $Y$ uniquely identifies outcomes,

$$
P(X=1 \mid Y=3)=1,\quad P(X=1 \mid Y=4)=1.
$$

Now,

$$
P(Y=3 \mid X=1) = \frac{0.10}{0.30} = \frac{1}{3},
$$

$$
P(Y=4 \mid X=1) = \frac{0.20}{0.30} = \frac{2}{3}.
$$


### 9. Interpretation

Conditioning on $Y$ gives complete information about the item, so there is no uncertainty about $X$.

However, conditioning on $X$ leaves uncertainty about which machine produced the item.

### 10. Information in $\sigma(X)$ vs $\sigma(Y)$

The σ-algebra $\sigma(X)$ only captures whether the item is:
- good, or
- defective.

It does not distinguish between machines.

In contrast, $\sigma(Y)$ contains full information about both quality and machine. Therefore, it is strictly finer than $\sigma(X)$.

##12. Shannon Entropy (Natural Logarithm)

The entropy formula is

$$
H(Z) = -\sum_z P(z)\ln P(z).
$$


### Entropy of $X$

$$
H(X) = -(0.70\ln 0.70 + 0.30\ln 0.30) \approx 0.61086.
$$


### Entropy of $Y$

$$
H(Y) = -(0.50\ln 0.50 + 0.20\ln 0.20 + 0.10\ln 0.10 + 0.20\ln 0.20)
\approx 1.22061.
$$


### Joint Entropy

$$
H(X,Y) = H(Y) \approx 1.22061.
$$


### Conditional Entropies

$$
H(X \mid Y) = 0,
$$

$$
H(Y \mid X) = H(X,Y) - H(X) \approx 0.60975.
$$


### Final Interpretation

- $H(X)$ measures uncertainty about whether an item is good or defective.
- $H(Y)$ measures uncertainty about the full classification (quality + machine).
- $H(X \mid Y) = 0$ because $Y$ completely determines $X$.
- $H(Y \mid X) > 0$ because $X$ does not determine which machine produced the item.

Therefore, $Y$ contains strictly more information than $X$, which is consistent with

$$
\sigma(X) \subseteq \sigma(Y).
$$

## Q10) Information Available in Continuous Random Varaibles

### 1. Probability Space

The probability space is:

$$
\Omega = \mathbb{R}^2, \qquad \mathfrak{F} = \mathfrak{B}(\mathbb{R}^2)
$$

A sample point is:
$$
\omega = (t,p)
$$

where:
- $t$ = standardized temperature deviation  
- $p$ = standardized pressure deviation  

The probability measure is defined by the joint density:

$$
f_{T,P}(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

For any Borel set $A \subseteq \mathbb{R}^2$:

$$
P(A) = \iint_A f_{T,P}(t,p)\, dt\, dp
$$

This defines a valid probability measure since the density integrates to 1.
<br>
### 2. $\sigma(X)$

The random variable is defined by:

$$
X(t,p) = t
$$

Thus, $X$ records only the temperature deviation.

The $\sigma$-algebra generated by $X$ is:

$$
\sigma(X) = \{ X^{-1}(B) : B \in \mathfrak{B}(\mathbb{R}) \}
$$

Since

$$
X^{-1}(B) = \{(t,p) \in \mathbb{R}^2 : t \in B\} = B \times \mathbb{R},
$$

we obtain:

$$
\sigma(X) = \{ B \times \mathbb{R} : B \in \mathfrak{B}(\mathbb{R}) \}
$$

These sets correspond to **vertical strips** in $\mathbb{R}^2$, since they depend only on the $t$-coordinate.
<br>
### 3. $\sigma(Y)$

The random variable is defined by:

$$
Y(t,p) = (t,p)
$$

Thus, $Y$ records both temperature and pressure.

The $\sigma$-algebra generated by $Y$ is:

$$
\sigma(Y) = \{ Y^{-1}(A) : A \in \mathfrak{B}(\mathbb{R}^2) \}
$$

Since $Y$ is the identity map on $\mathbb{R}^2$, we have:

$$
Y^{-1}(A) = A
$$

for every Borel set $A \subseteq \mathbb{R}^2$.

Hence,

$$
\sigma(Y) = \mathfrak{B}(\mathbb{R}^2)
$$

So $\sigma(Y)$ contains **all Borel subsets of $\mathbb{R}^2$**, meaning it captures complete information about both variables.
<br>
### 4. Relation between $\sigma(X)$ and $\sigma(Y)$

We have:

$$
\sigma(X) \subseteq \sigma(Y)
$$

This follows because:
- $X(t,p) = t$ depends only on the temperature coordinate,
- $Y(t,p) = (t,p)$ contains both temperature and pressure information.

More formally, every set in $\sigma(X)$ has the form:

$$
B \times \mathbb{R}
$$

and since $Y$ generates all Borel subsets of $\mathbb{R}^2$, we have:

$$
B \times \mathbb{R} \in \mathfrak{B}(\mathbb{R}^2) = \sigma(Y)
$$

Hence, $\sigma(X)$ is a sub-$\sigma$-algebra of $\sigma(Y)$.

**Interpretation:**  
$Y$ contains strictly more information than $X$, since $Y$ observes both temperature and pressure, while $X$ observes only temperature.




<br>

### 5. Marginal Density of $X$

The marginal density of $X$ is obtained by integrating out $p$:

$$
f_X(x) = \int_{-\infty}^{\infty} f_{T,P}(x,p)\, dp
$$

Substituting the joint density:

$$
f_X(x) = \int_{-\infty}^{\infty} \frac{1}{4\pi} \exp\left(-\frac{x^2}{2} - \frac{p^2}{8}\right) dp
$$

Factorizing terms independent of $p$:

$$
f_X(x) = \frac{1}{4\pi} e^{-x^2/2} \int_{-\infty}^{\infty} e^{-p^2/8} dp
$$

Using the Gaussian integral:

$$
\int_{-\infty}^{\infty} e^{-p^2/8} dp = \sqrt{8\pi}
$$

We obtain:

$$
f_X(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}
$$

Hence,

$$
X \sim N(0,1)
$$
<br>

### 6. Marginal Density of $Y$

Since $Y = (T, P)$ is the full random vector, its distribution is already given by the joint density:

$$
f_Y(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

Thus, $Y$ is a bivariate normal random vector with independent components:
- $T \sim N(0,1)$
- $P \sim N(0,4)$
<br>
### 7. Probabilities

We compute the required probabilities using properties of the normal distribution and independence.

     

**1. Probability that $X \le 0$**

Since $X \sim N(0,1)$:

$$
P(X \le 0) = \frac{1}{2}
$$


**2. Probability that $X > 1$**

Using the standard normal CDF $\Phi$:

$$
P(X > 1) = 1 - \Phi(1) \approx 0.1587
$$



**3. Probability for $Y$ in a half-plane**

$$
P\big(Y \in (-\infty,0] \times \mathbb{R}\big)
= P(T \le 0)
= \frac{1}{2}
$$



**4. Probability for rectangular region**

Using independence of $T$ and $P$:

$$
P\big(Y \in [-1,1] \times [-2,2]\big)
= P(-1 \le T \le 1)\, P(-2 \le P \le 2)
$$

Now,

$$
P(-1 \le T \le 1) = 2\Phi(1) - 1
$$

and since $P \sim N(0,4)$:

$$
P(-2 \le P \le 2) = 2\Phi(1) - 1
$$

Therefore,

$$
P\big(Y \in [-1,1] \times [-2,2]\big) = (2\Phi(1)-1)^2 \approx 0.466
$$
<br>
### 8. Conditional Distribution of $X \mid Y$

Since $X(t,p) = t$ and $Y(t,p) = (t,p)$, the value of $X$ is completely determined once $Y$ is known.

Thus, the conditional distribution of $X$ given $Y=(t,p)$ is degenerate:

$$
X \mid Y = (t,p) \sim \delta_t
$$

where $\delta_t$ is the Dirac measure concentrated at $t$.

Equivalently, for any Borel set $A \subseteq \mathbb{R}$:

$$
P(X \in A \mid Y=(t,p)) = \mathbf{1}_A(t)
$$

This means:
- If $t \in A$, probability is 1  
- If $t \notin A$, probability is 0  

**Interpretation:**  
Once both temperature and pressure are known, the temperature is known exactly with no uncertainty.
<br>
### 9. Conditional Distribution of $Y \mid X=t$

The conditional distribution of $Y$ given $X=t$ is obtained by noting that
$X(t,p)=t$ fixes the temperature component, while the pressure component remains random.

Thus:

$$
Y \mid X=t = (t, P)
$$

where:

$$
P \sim N(0,4)
$$



#### Measure-theoretic form

Since $T$ and $P$ are independent, the conditional probability measure is:

$$
P_{Y|X=t} = \delta_t \otimes \mathcal{N}(0,4)
$$

where $\delta_t$ is the Dirac measure at $t$.



#### Conditional density (formal form)

The conditional density with respect to Lebesgue measure is:

$$
f_{Y|X}\big((u,v)\mid t\big)
= \delta_t(u)\cdot \frac{1}{\sqrt{8\pi}} \exp\left(-\frac{v^2}{8}\right)
$$



### Interpretation

- The temperature component is fixed at $t$ (no randomness → Dirac delta)
- The pressure component remains Gaussian with variance $4$
- Hence only pressure uncertainty remains after conditioning on $X$
<br>
### 10. Interpretation

- **Conditioning on $X$:**  
  We only know the temperature deviation. The pressure deviation remains uncertain and follows its original distribution.

- **Conditioning on $Y$:**  
  We know both temperature and pressure deviations exactly. There is no remaining uncertainty about either variable.

**Conclusion:**  
Conditioning on $Y$ provides strictly more information than conditioning on $X$, since $Y$ fully determines the state $(T, P)$.
<br>
### 11. Measurability

Define the projection map:

$$
\pi_1 : \mathbb{R}^2 \to \mathbb{R}, \qquad \pi_1(t,p) = t
$$

Since:

$$
Y(t,p) = (t,p)
$$

we have:

$$
X = \pi_1 \circ Y
$$

i.e.,

$$
X(t,p) = \pi_1(Y(t,p))
$$



### Conclusion

Since $\pi_1$ is a Borel-measurable function and $Y$ is measurable, it follows that $X$ is measurable with respect to $\sigma(Y)$.

Thus:

$$
\sigma(X) \subseteq \sigma(Y)
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu_T, sigma_T = 0.0, 1.0
mu_P, sigma_P = 0.0, 2.0   # Var(P)=4


# Grids

t = np.linspace(-4, 4, 400)
p = np.linspace(-8, 8, 400)
T, P = np.meshgrid(t, p)


# Joint density f(T,P)

joint_pdf = (
    norm.pdf(T, loc=mu_T, scale=sigma_T) *
    norm.pdf(P, loc=mu_P, scale=sigma_P)
)


# Marginals

fx = norm.pdf(t, loc=mu_T, scale=sigma_T)
fp = norm.pdf(p, loc=mu_P, scale=sigma_P)


# Conditional P | X=t0

t0 = 1.0
cond_p_given_x = fp  # independence ⇒ same distribution


# Conditional X | Y=(t1,p1)
# (visual approximation of Dirac delta)

t1, p1 = 1.0, 2.0
eps = 0.07
cond_x_given_y_vis = norm.pdf(t, loc=t1, scale=eps)


# Figure layout

fig = plt.figure(figsize=(12, 11))
gs = fig.add_gridspec(3, 2, hspace=0.45, wspace=0.3)


# (1) Joint density

ax1 = fig.add_subplot(gs[0, :])
cs = ax1.contourf(T, P, joint_pdf, levels=25, cmap="viridis")
ax1.set_title(r"Joint Density of $Y=(T,P)$")
ax1.set_xlabel("t (temperature)")
ax1.set_ylabel("p (pressure)")
fig.colorbar(cs, ax=ax1, label="density")


# (2) Marginal of X

ax2 = fig.add_subplot(gs[1, 0])
ax2.plot(t, fx, lw=2, color="navy")
ax2.fill_between(t, fx, alpha=0.3, color="skyblue")
ax2.set_title(r"Marginal of $X=T$")
ax2.set_xlabel("x")
ax2.set_ylabel("density")
ax2.grid(alpha=0.3)


# (3) Marginal of P

ax3 = fig.add_subplot(gs[1, 1])
ax3.plot(p, fp, lw=2, color="darkgreen")
ax3.fill_between(p, fp, alpha=0.3, color="lightgreen")
ax3.set_title(r"Marginal of $P$")
ax3.set_xlabel("p")
ax3.set_ylabel("density")
ax3.grid(alpha=0.3)


# (4) P | X = t0

ax4 = fig.add_subplot(gs[2, 0])
ax4.plot(p, cond_p_given_x, lw=2, color="crimson",
         label=r"$P|X=t_0$")
ax4.fill_between(p, cond_p_given_x, alpha=0.3, color="salmon")
ax4.set_title(r"Conditional $P \mid X=t_0$")
ax4.set_xlabel("p")
ax4.set_ylabel("density")
ax4.legend()
ax4.grid(alpha=0.3)


# (5) X | Y = (t1,p1)

ax5 = fig.add_subplot(gs[2, 1])
ax5.plot(t, cond_x_given_y_vis, lw=2, color="purple",
         label="approx. Dirac at t1")
ax5.axvline(t1, color="black", ls="--")
ax5.set_title(r"$X \mid Y=(t_1,p_1)$")
ax5.set_xlabel("x")
ax5.set_ylabel("density")
ax5.legend()
ax5.grid(alpha=0.3)

plt.show()
```

### 13. Differential Entropy of $X$

Since $X \sim N(0,1)$, the differential entropy is given by:

$$
h(X) = \frac{1}{2} \ln(2\pi e \sigma^2)
$$

Here, $\sigma^2 = 1$, so:

$$
h(X) = \frac{1}{2} \ln(2\pi e)
$$



### Numerical value

$$
h(X) \approx 1.4189
$$
<br>
### 14. Differential Entropy of $Y$

Since $Y = (T, P)$ and $T$ and $P$ are independent Gaussian random variables:

- $T \sim N(0,1)$  
- $P \sim N(0,4)$  

The differential entropy is:

$$
h(Y) = h(T) + h(P)
$$

For a Gaussian random variable $N(0,\sigma^2)$:

$$
h = \frac{1}{2} \ln(2\pi e \sigma^2)
$$

So:

$$
h(Y) = \frac{1}{2} \ln(2\pi e) + \frac{1}{2} \ln(8\pi e)
$$

This simplifies to:

$$
h(Y) = \ln(4\pi e)
$$



### Numerical value

$$
h(Y) \approx 3.5310
$$
<br>
### 15. Conditional Differential Entropy

Since $Y = (T, P)$ and $T$ and $P$ are independent, conditioning on $X = T$ leaves only the uncertainty in $P$.

Thus:

$$
h(Y|X) = h(P|T)
$$

But independence implies:

$$
h(P|T) = h(P)
$$

Since $P \sim N(0,4)$, we use the Gaussian entropy formula:

$$
h(P) = \frac{1}{2} \ln(2\pi e \cdot 4)
$$

So:

$$
h(Y|X) = \frac{1}{2} \ln(8\pi e)
$$



### Numerical value

$$
h(Y|X) \approx 2.1121
$$



### Interpretation

This represents the remaining uncertainty in pressure after the temperature is known.
<br>
### 16. Important Result: $h(X|Y)$

Since $X = \pi_1(Y)$, the random variable $X$ is completely determined by $Y$.

Thus, the conditional distribution of $X$ given $Y$ is a Dirac measure:

$$
X \mid Y = \delta_{X(Y)}
$$



### Result

In the continuous case, this leads to:

$$
h(X|Y) = -\infty
$$



### Reason

- $X$ is a deterministic function of $Y$
- The conditional distribution collapses to a point mass (Dirac delta)
- Dirac measures are not absolutely continuous
- Differential entropy is not well-defined in the usual sense and tends to $-\infty$


### Key Insight

Unlike the discrete case where:
$$
H(X|Y) = 0
$$

In the continuous case:
$$
h(X|Y) = -\infty
$$

This highlights a fundamental difference between discrete and continuous entropy.
